## 3. Set API keys

```python
import os
from getpass import getpass

def load_secret(name, required):
    try:
        from google.colab import userdata
        value = userdata.get(name)
    except Exception:
        value = None
    if not value and required:
        value = getpass(f"{name}: ")
    if value:
        os.environ[name] = value
    print(f"{name}: {'set' if value else 'not set'}")

load_secret("OPENAI_API_KEY", required=True)
load_secret("S2_API_KEY", required=False)
```
